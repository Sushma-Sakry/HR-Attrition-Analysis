```python
# ==========================================
# Import Libraries
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# Load Dataset
# ==========================================

df = pd.read_csv("hr_employee_attrition.csv")

# ==========================================
# Display First 5 Rows
# ==========================================

df.head()
```

```python
# Shape of the dataset
print("Shape of dataset:", df.shape)

# Number of rows
print("Number of rows:", df.shape[0])

# Number of columns
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
print("Duplicated rows", df.duplicated().sum())
```

```python
# ==========================================
# Drop Unnecessary Columns
# ==========================================

df = df.drop(
    columns=[
        "EmployeeCount",
        "Over18",
        "StandardHours"
    ]
)

print(df.shape)
```

```python
df.columns
```

```python
# ==========================================
# Attrition Distribution
# ==========================================

import matplotlib.pyplot as plt

attrition_counts = df["Attrition"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    attrition_counts.index,
    attrition_counts.values
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.tight_layout()

plt.show()
```

```python
# Attrition Percentage

attrition_percentage = (
    df["Attrition"]
    .value_counts(normalize=True)
    * 100
)

print(attrition_percentage)
```

## Business Insight

- The dataset contains 1,470 employee records.
- Approximately **83.88%** of employees remained with the company.
- Around **16.12%** of employees left the organization.
- The dataset is moderately imbalanced, with attrition representing a smaller proportion of the workforce.
- Understanding the factors driving this 16.12% attrition will be the primary focus of the analysis.

# Gender Distribution Analysis

```python
# Gender Distribution

gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    gender_counts.index,
    gender_counts.values
)

plt.title("Gender Distribution of Employees")
plt.xlabel("Gender")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

print(gender_counts)
```

```python
# Attrition by Gender

attrition_gender = (
    df[df["Attrition"] == "Yes"]
    ["Gender"]
    .value_counts()
)

plt.figure(figsize=(8,5))

plt.bar(
    attrition_gender.index,
    attrition_gender.values
)

plt.title("Employee Attrition by Gender")
plt.xlabel("Gender")
plt.ylabel("Employees Left")

plt.tight_layout()
plt.show()

print(attrition_gender)
```

```python
# Attrition Rate by Gender

gender_summary = (
    df.groupby("Gender")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
)

print(gender_summary)
```

## Business Insight

- The workforce consists of 882 male employees and 588 female employees.
- Male employees account for the majority of the workforce.
- Male employees have a slightly higher attrition rate (17.01%) compared to female employees (14.80%).
- The difference is relatively small, suggesting gender is not a primary factor influencing employee attrition.


# Department Analysis

```python
department_counts = df["Department"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    department_counts.index,
    department_counts.values
)

plt.title("Employee Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()

print(department_counts)
```

```python
department_attrition = (
    df[df["Attrition"] == "Yes"]
    ["Department"]
    .value_counts()
)

plt.figure(figsize=(8,5))

plt.bar(
    department_attrition.index,
    department_attrition.values
)

plt.title("Attrition Count by Department")
plt.xlabel("Department")
plt.ylabel("Employees Left")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()

print(department_attrition)
```

```python
department_attrition_rate = (
    df.groupby("Department")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

plt.bar(
    department_attrition_rate.index,
    department_attrition_rate.values
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()

print(department_attrition_rate)
```

## Business Insight

- Research & Development is the largest department with 961 employees.
- Sales has the highest attrition rate at 20.63%.
- Human Resources has the second-highest attrition rate at 19.05%.
- Research & Development shows the lowest attrition rate at 13.84%.
- Employee retention efforts should prioritize the Sales department, where employee turnover is most severe.

# Age Analysis

```python
# Age Distribution

plt.figure(figsize=(8,5))

plt.hist(
    df["Age"],
    bins=10
)

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
# Create Age Groups

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18,25,35,45,55,60],
    labels=["18-25","26-35","36-45","46-55","56+"],
    include_lowest=True
)

df["AgeGroup"].value_counts()
```

```python
age_group_counts = df["AgeGroup"].value_counts()

plt.figure(figsize=(8,5))

plt.bar(
    age_group_counts.index,
    age_group_counts.values
)

plt.title("Employee Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()

print(age_group_counts)
```

```python
age_attrition = (
    df.groupby("AgeGroup")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))

plt.bar(
    age_attrition.index,
    age_attrition.values
)

plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")

plt.tight_layout()
plt.show()

print(age_attrition)
```

## Business Insight

- The majority of employees belong to the 26-35 and 36-45 age groups.
- Employees aged 18-25 exhibit the highest attrition rate at 35.77%.
- Employees aged 36-45 show the lowest attrition rate at 9.19%.
- Attrition generally decreases as employees gain age and experience.
- The organization should focus on retaining younger employees through improved onboarding, career development programs, and mentorship initiatives.

# Final Conclusions

## Key Findings

### 1. Overtime is the strongest attrition driver

- Employees working overtime show an attrition rate of 30.53%.
- Employees not working overtime show an attrition rate of 10.44%.
- Overtime employees are nearly three times more likely to leave.

### 2. Younger employees are more likely to leave

- Employees aged 18–25 have the highest attrition rate at 35.77%.
- Attrition decreases as employee age and experience increase.

### 3. Frequent business travel increases attrition

- Employees who travel frequently have an attrition rate of 24.91%.
- Non-travel employees show only 8.00% attrition.

### 4. Salary impacts retention

- Low salary employees show an attrition rate of 28.61%.
- High salary employees show only 10.80% attrition.

### 5. Certain job roles are at higher risk

- Sales Representatives have the highest attrition rate at 39.76%.
- Laboratory Technicians show an attrition rate of 23.94%.

### 6. Sales Department needs attention

- Sales department has the highest attrition rate at 20.63%.

## Recommendations

- Reduce excessive overtime workload.
- Improve compensation and career growth opportunities for lower-paid employees.
- Strengthen onboarding and mentorship programs for younger employees.
- Provide additional support for frequent travelers.
- Focus retention strategies on Sales Representatives and Laboratory Technicians.
